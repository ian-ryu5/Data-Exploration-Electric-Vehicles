### Dataset Narrative: Electric Vehicle Population Data Analysis

#### Dataset Selection
The selected dataset provides a comprehensive view of electric vehicles (EVs) registered in various counties. It includes fields such as vehicle make, model, model year, electric vehicle type, electric range, base MSRP, and geographic information (county, city, and postal code). The dataset is valuable for examining the spread of EV adoption, preferences for certain vehicle types, and trends in electric vehicle range across regions. It can help provide insights into how EV usage patterns vary by location and type, potentially informing policy, infrastructure, and consumer behavior.

#### Data Cleaning and Exploration
To prepare the data for analysis, I first examined missing values and data types. Key columns like “Electric Range” had some missing values, and we filtered these out for visual consistency. Additionally, data was grouped by county and electric vehicle type to calculate the average electric range, which enabled meaningful comparisons across counties and vehicle types. Outliers, such as unrealistically high or zero ranges, were identified and excluded, assuming that these might represent data errors.

#### Data Visualization Process
I focused on visualizing the average electric range by county and electric vehicle type (Battery Electric Vehicle vs. Plug-in Hybrid). Using ggplot2, I created a bar chart with county names on the x-axis and average electric range on the y-axis. This visualization highlights the variation in EV adoption by region and by vehicle type, as well as the prevalence of longer-range BEVs in certain counties. The choice of contrasting colors for BEVs and PHEVs helps clearly differentiate the types, revealing regional preferences.

#### Findings and Conclusions
The visualization reveals that certain counties favor BEVs with higher ranges, which could be due to urban infrastructure that supports long-range EVs, or consumer preferences in those areas. Conversely, counties with lower average ranges or a higher proportion of PHEVs may lack EV charging infrastructure or have more rural areas where PHEVs are preferred for their extended range using fuel. This insight can be valuable for local governments or businesses planning to invest in EV infrastructure.

#### Dataset Limitations
Despite its insights, this dataset has limitations. First, it lacks temporal data that would allow for trend analysis of EV adoption over time. Second, the dataset does not include demographic information, which would be helpful to assess the impact of income or population density on EV adoption. Finally, while base MSRP is included, more detailed cost information could further clarify economic factors influencing EV adoption. Adding these variables would enhance the analysis and enable a more comprehensive understanding of the factors influencing EV trends by county.

```r
avg_range_by_maker <- ev_data %>%
  filter(!is.na(`Electric Range`)) %>%
  group_by(Make) %>%
  summarise(avg_range = mean(`Electric Range`, na.rm = TRUE)) %>%
  arrange(desc(avg_range)) %>%
  slice_max(order_by = avg_range, n = 10)  # Select top 10 car makers by range

ggplot(avg_range_by_maker, aes(x = reorder(Make, avg_range), y = avg_range, fill = Make)) +
  geom_bar(stat = "identity") +
  coord_flip() +
  labs(
    title = "Top 10 Car Makers by Average Electric Range",
    x = "Car Maker",
    y = "Average Electric Range (miles)"
  ) +
  theme_minimal() +
  theme(legend.position = "none") +
  scale_fill_viridis_d(option = "plasma")
```
